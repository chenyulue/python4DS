## 简介

我们将协助您完成以下配置：

后续章节将提供更详细的操作指南，在此我们首先解释这些概念的定义。

### 集成开发环境（IDE）

*集成开发环境*（IDE）是一种提供多种编码辅助工具的软件应用程序。其核心功能是为代码编写提供载体！虽然IDE并非编程的唯一方式，但可能是最高效的选择。需注意，编程语言与代码编写载体（IDE）是相互独立的：语言是执行指令的规范，而IDE是书写指令的场所。市面上存在众多集成开发环境，本书强烈推荐微软开发的*Visual Studio Code*。该IDE兼容所有主流操作系统，且为当前最流行的工具之一。以下是*Visual Studio Code*提供的实用功能：
- 支持交互式逐行运行代码或一次性执行代码
- 提供代码调试（错误排查）功能
- 快速访问常用软件包的帮助文档
- 自动代码格式化，确保代码符合最佳实践规范
- 通过<kbd>TAB</kbd>键实现代码自动补全
- 基础语法错误自动检查
- 括号配对高亮显示，便于追踪代码逻辑执行顺序

### Python解释器

Python既是一种人类可读的编程语言，也是计算机可解析、解释并执行指令的机器语言。为使计算机能够读取和执行Python代码，需在本地安装Python解释器。尽管存在多种Python解释器安装方式，但本书推荐使用具备灵活性与简易性优势的*uv发行版*。

### 软件包

Python**软件包**是扩展Python功能的函数、数据及文档的集合。由于数据科学所需的核心功能大多依赖扩展包实现，因此软件包的使用至关重要。在Python脚本开头常见`import numpy as np`等语句——这是调用已安装软件包（此处为`numpy`包）并为后续代码使用该软件包指定其简化名称（此处为`np`）的指令。通过`np.`语法即可访问`numpy`包中的函数，例如使用`np.log(x)`计算变量`x`的対数值（假设`x`为数值型变量）。软件包仅需安装一次即可永久使用。

### 典型工作流程

基于代码的数据分析通常遵循以下工作流程：
1. 启动*集成开发环境*（IDE）
2. 在IDE中编写脚本代码（含代码的文本文件）
3. 根据分析需求安装必要的扩展*软件包*
4. 通过IDE将脚本中的部分代码或完整脚本发送至*Python解释器*及扩展*软件包*执行，并展示运行结果

我们将介绍两种实现该工作流程的方式：
1. 在本地计算机安装*IDE*、*Python解释器*及扩展*软件包*
2. 使用基于互联网浏览器访问的云端计算机。该云端环境已预装*IDE*和*Python解释器*，并可便捷安装扩展*软件包*，但免费版本每月仅限使用60小时

建议选择您更习惯的操作方式！未来您可能会尝试两种方案。

### 安装Python

为下载并安装Python，我们将采用适用于所有主流操作系统的uv “Python 发行版”。具体安装步骤请参照[此网站](https://docs.astral.sh/uv/getting-started/installation/#installation-methods)的说明。与常规软件安装不同，此处需通过*命令行*执行Python安装操作。Linux、Mac及Windows系统均内置命令行工具：Mac或Linux系统搜索*Terminal*，Windows系统搜索*Powershell*。启动对应应用程序后，将出现可输入命令的交互界面。截至本文撰写时，安装命令如下：

适用于Linux与Mac系统的命令为：

```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

而适用于Windows系统的命令为：

```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

输入命令后按回车键执行。

完成uv安装后，可通过执行`uv --version`验证其安装状态。若终端显示包含"uv"标识及当前版本号的信息，则表明安装成功。

需特别说明的是，此时尚未实际安装Python——我们安装的是用于管理Python环境的**uv工具**。因此，下一步需执行以下命令：

```bash
uv python install
```

保存文件。

现在正式**运行代码**：选中`print("Hello world!")`行，右键单击选择**"在交互式窗口中运行选中行"**。

*交互式窗口*支持灵活运行脚本代码或直接输入代码指令。该窗口会持久保存所有变量状态（如`x = 5`这类赋值语句），无论变量源自脚本运行或窗口直接输入。熟悉Stata、Matlab或R的用户将感到亲切——无需预先编写完整脚本，可逐行调试、实时修改并重新执行。

为避免频繁右键操作，我们创建*键盘快捷键*实现代码快速执行：

- 打开VS Code配置菜单（左下角齿轮图标）
- 进入"设置"
- 搜索"jupyter send"定位"Interactive Window > Text Editor: Execute Selection"选项
- 确保该选项处于勾选状态

返回脚本文件，光标定位至`print("Hello world!")`行，按下<kbd>Shift</kbd>+<kbd>Enter</kbd>组合键，即可再次输出"Hello world!"，操作效率显著提升。

随后按下回车键。其中`包名称`可替换为具体包名，例如`pandas`。若遇安装问题，请确保网络连接正常，且[PyPI](https://pypi.org/)（Python官方软件包索引）未被防火墙或代理拦截。以安装数据分析包**polars**为例，可执行`uv add polars`命令。后续章节将详细介绍**polars**使用方法，若需验证安装是否成功，只需查看终端输出的"Successfully installed polars"提示信息及对应版本号。

通过终端输入`uv pip list`指令，可查看当前环境中已安装的所有软件包及其版本信息。

关于软件包安装与使用的进阶内容，详见{ref}`workflow-packages-and-environments`章节。

若安装过程出现问题，请再次确认网络连接状态，并检查防火墙或代理设置是否对[PyPI](https://pypi.org/)访问造成了阻碍。
