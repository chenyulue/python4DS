```python
# remove-input
import graphviz

dot = graphviz.Digraph(comment="Data science workflow")
dot.attr(compound="true")
dot.edge("导入", "清洗")

with dot.subgraph(name="cluster_0") as c:
    c.attr(style="filled", color="lightgrey")
    c.node_attr.update(style="filled", color="white")
    c.edges(
        [("可视化", "分析"), ("分析", "转换"), ("转换", "可视化")]
    )
    c.attr(label="理解")

dot.edge("清洗", "分析", lhead="cluster_0")

dot.edge("分析", "交流", ltail="cluster_0")

dot
```

首先必须将数据**导入**Python环境。这通常意味着将存储在文件、数据库或Web应用程序编程接口（API）中的数据加载到Python的"数据框（data frame）"中。没有数据，数据科学就无从谈起！

数据导入后，通常需要进行**清洗**或整理。数据清洗是指将数据以一致的格式存储，使数据集的语义与其存储方式相匹配。简而言之，就是让数据格式更易于操作！所谓"整洁"数据是指结构化表格数据的特例形式：每列代表一个变量，每行代表一个观测值（但并非所有数据都是表格形式）。清洗后的数据至关重要，因为统一的结构能让我们专注于关于数据的问题的解答，而非耗费精力调整数据格式以适应不同函数。

获得整洁数据后，下一步通常是**转换**操作。数据转换包括筛选目标观测值（如筛选某城市的所有居民或最近一年的数据）、基于现有变量创建新变量（如通过距离和时间计算速度），以及生成汇总统计量（如计数或均值）。

当获得包含所需变量的整洁数据后，知识生成主要依靠两大引擎：可视化与分析。二者优劣互补，因此实际分析过程中需要多次交替使用。

**可视化**本质上是一项以人为本的活动。优秀的可视化能揭示意外发现，或引发新的关于数据的问题；它还可能暗示当前询问的问题不恰当，或需要采集其他数据。可视化往往会以多种多样的表现形式出人意料！

**分析**涉及面过于广泛，本书难以全面涵盖。它可能包括建立模型、进行统计、解答具体问题或从数据中提炼叙事逻辑。这是项重要技能，但高度依赖具体应用领域。

数据科学的最后环节是**成果沟通**，这是任何数据分析项目的关键环节。即使通过模型和可视化获得了深刻洞见，若无法向他人有效传达研究成果，所有努力都将失去价值。

贯穿全过程的工具是**编程**。编程作为通用工具，几乎应用于数据科学项目的每个环节。要成为优秀的数据科学家，虽无需成为编程专家，但提升编程技能将事半功倍——更强的编程能力有助于自动化常规任务，更从容地解决新问题。

这些工具适用于所有数据科学项目，但对多数项目而言，仅有这些仍显不足。这里存在粗略的80/20法则：使用本书教授的工具可完成约80%的项目内容，剩余20%需要借助其他工具。本书将为您指引深入学习这些工具的资源路径。

## 大数据处理

本书聚焦于"内存驻留"型小数据集处理（即普通笔记本电脑可直接打开的数据规模）。这是理想的入门起点，因为驾驭大数据的前提是精通小数据处理。本书所授工具可轻松处理数百MB级数据，通过适当优化甚至可应对1-2GB数据量。如需常规处理10-100GB级数据，建议深入学习数据库技术及[Ibis](https://ibis-project.org/)等大数据交互工具。

（若数据规模远超此范围，建议审慎评估问题本质：看似庞大的数据集，其具体分析目标所需数据可能仍是小规模。通过数据子集提取、抽样或特征聚合等策略，往往能在内存限制下实现有效分析。此过程的核心挑战在于精准定位关键数据，通常需要大量迭代优化。）

另一种可能是：所谓大数据问题实为海量小数据任务的集合。例如需为百万用户分别建立预测模型，当单个模型构建可在内存完成时，只需借助[Hadoop](https://hadoop.apache.org/)或[Spark](https://spark.apache.org/)等分布式系统实现高度并行计算。掌握本书方法后，可进一步学习**pyspark**等工具实现全量数据处理。
